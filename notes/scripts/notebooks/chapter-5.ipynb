# 📘 Modules and Files

## 🚀 Introduction
- A module is a .py file containing Python definitions and statements.


````python
import circle
pi = 3
print(pi)
print(circle.pi)
print(circle.area(3))
print(circle.sphere_surface(4))
````

- Modules are typically stored in individual files. 

- Each module has its own private symbol table. 

-  Executing `import M` creates a binding for module `M` in the scope in which the `import`
appears. 

- In the importing context we use dot notation to indicate that we are referring to a name defined in the imported module

- The use of dot notation to fully qualify names avoids the possibility of getting burned by an accidental name clash

- A module can contain executable statements as well as function definitions

- The statements are used to initialize the module

- The statements in a module are executed only the first time a module is imported into a program

- A module is imported only once per interpreter session

- A variant of the `import` statement that allows the importing program to omit the module name when accessing names defined inside the imported module

- `from m import *` creates bindings in the current scope to all objects defined within `M`, but not to `M` itself

````python
from circle import *
print(pi)
print(circle.pi)
````

- Many Python programmers frown upon using this kind of “wild card” import. They believe that it makes code more difficult to read because it is no longer obvious where a name (for example pi in the above code) is defined.

- `import module_name as new_name`

- The most common reason programmers use this form is to provide an abbreviation for a long name

````python
import circle as my_circle
````

#### 🌀  Import Statements Are Executable

Many people think of `import` as just a way to bring in functions or classes, like in a static way. But actually:
- Import is executable code.

- When Python reaches an import statement during execution, it runs the module’s code — not just loads it.

- That means the moment you import something, the module might do stuff, not just sit there.

#### 🌀 Modules Can Run Code When Imported

- Some modules contain top-level code that runs immediately when imported

#### 🌀 Potential Problem: Shared Resources

Let’s say a module you import:
- Opens a file.
- Starts a network connection.
- Accesses a database.
- Allocates some hardware resource.

These are shared system resources, and they could be sensitive to when or how often they are accessed.

So, if you import such a module:
- Too early → maybe the resource isn’t ready yet.
- Too late → maybe you needed it already.
- Multiple times (through reloads) → maybe the module re-initializes the resource in an unsafe way. 

```python
# logging_module.py
with open("app.log", "a") as f:
    f.write("Log file initialized\n")
```

### 🌀 Why Most People Don’t Face This

In most programs:
- The standard Python libraries you import (like math, datetime, os, etc.) do not do anything dangerous or resource-heavy at import time.
- They just define functions/classes and maybe set up a bit of memory.

So in most cases, it’s totally fine to put all your imports at the top — and that’s the recommended practice in Python style guides like PEP 8.

### 🌀 When to Delay an Import

You might want to import something later in the code if:
- You only need it under specific conditions (e.g., inside a function).
- It’s slow or resource-heavy to import.
- You want to avoid circular dependencies between modules.
- You want to control when that top-level initialization code in the module runs.




# 📘 Files 

- Every computer system uses files to save things from one computation to the next

- Each operating system (e.g., Windows and macOS) comes with its own file system for creating and accessing files

- Python achieves operating-system independence by accessing files through something called a file handle

```python
name_handle = open('kids', 'w')
```

- instructs the operating system to create a file with the name kids and return a file handle for that file


````python
name_handle = open('kids', 'w')
for i in range(2):
    name = input('Enter name: ')
    name_handle.write(name + '\n')
name_handle.close()
````

- can ensure that you don't forget to close a file by opening it using a with statement. 
- Code of the form
```python
with open(file_name) as name_handle:
    code_block
```
- Opens a file, binds a local name to it that can be used in the code_block, and then closes the file when code_block is exited.

````python
with open('kids', 'r') as name_handle:
    for line in name_handle:
        print(line)
````

**🤔 What is happening**

**🪜 Step 1:** `with open(...) as f`

This is a context manager. What it does:
- Opens the file
- Ensures the file is automatically closed when the block ends — even if an error happens
- Assigns the file object to variable f 
    ```python
    with open("file.txt", "r") as f: 
    ``` 
    is like 


```python
f = open("file.txt", "r")
try:
        # do stuff
finally:
        f.close()
```

**🪜 Step 2:** `for line in f`

The file object f is an iterator over the lines of the file.

That means:
- When you loop over f, you’re reading the file line-by-line
- Internally, Python uses the `__iter__()` and` __next__()` methods
- It does not read the whole file into memory
- This 
    ```python
    for line in f:
        print(line)
    ```
 behaves like 

```python 
    while True:
       try:
           line = next(f)  # get the next line
       except StopIteration:
           break
       print(line)
```


- If line comes from a file, it usually ends with a newline (\n), so `print(line)` will add another newline, resulting in a blank line between each print.

````python
line = "hello\n"
print(line)
print('hello')
````

-  To avoid **overwritten** we can open the file for **appending** (instead of writing) by using the argument `'a'`
