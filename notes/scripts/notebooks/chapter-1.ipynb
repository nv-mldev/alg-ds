**Exercise-1:** Write a program that examines three variables—x, y, and z—and prints the largest odd number among them. If none of them are odd, it should print the smallest value of the three.
> Paradigm : Iterative accumulation pattern
1. Initializing a variable (answer).
2. Iterating over a sequence.
3. Applying a conditional check
4. Updating answer when necessary.
5. Printing the final result with a single print statement.

````python
x,y,z = (10,15,3)
#default answer : smallest value of 3
answer = min(x,y,z)
if x % 2 == 1 and x > answer:
    answer = x 
if y % 2 == 1 and y > answer:
    answer = y 
if z % 2 == 1 and z > answer:
    answer = z 
print(answer)
    
````

#### Conditional Expressions

Python supports conditional expressions as well as conditional statements. Conditional expressions are of the form
> expr1 if condition else expr2

If the condition evaluates to True, the value of the entire expression is expr1; otherwise it is expr2. For example, the statement
> x = y if y > z else z

sets x to the maximum of y and z.

````python
# find maximum using conditional expression 

y = 5 
z = 15
x = y if y > z else z 
print(x)
````

*A conditional expression can appear any place an ordinary expression can appear, including within conditional expressions*. 

Below example showcase the maximum of 3 numbers 

````python
(x,y,z) = (20,30,35)
print( (x if x > y else y ) if x > z else (z if z > y else y) )
````


* The class of branching programs is still quite limited.
* One way to think about the power of a class of programs is in terms of how long they can take to run. Assume that each line of code takes one unit of time to execute. If a straight-line program has n lines of code, it will take n units of time to run. 

* What about a branching program with n lines of code? 
    * It might take less than n units of time to run, but it cannot take more, since each line of code is executed at most once.

* A program for which the maximum running time is bounded by the length of the program is said to run in **constant time**. This does not mean that each time the program is run it executes the same number of steps. It means that there exists a constant, k, such that the program is guaranteed to take no more than k steps to run. This implies that the running time does not grow with the size of the input to the program.

### 📌 Character Encoding in Python: ASCII vs Unicode vs UTF-8
#### 🔹 What is Character Encoding?
Character encoding is a way to represent text in binary format so that computers can store and process it. Python supports various encodings, but the most important ones are:

1. **ASCII** (American Standard Code for Information Interchange)
2. **Unicode** (Universal Character Set)
3. **UTF-8** (Unicode Transformation Format - 8 bit)

---

##### **1️⃣ ASCII (American Standard Code for Information Interchange)**
- **Bit Size**: 7-bit (0-127)
- **Characters Supported**: English letters (A-Z, a-z), digits (0-9), punctuation, and control characters.
- **Limitation**: Cannot represent non-English characters

````python
char = 'A'
ascii_code = ord(char)
print(ascii_code)

ascii_code = 65
char = chr(ascii_code)
print(char)
````

---

##### **2️⃣ Unicode (Universal Character Set)**
- **Bit Size**: Variable (8, 16, or 32-bit)
- **Supports**: All known languages, symbols, and emojis.
- **Backward Compatible with ASCII**: The first 128 Unicode characters match ASCII.

````python
char = '字'
unicode_code = ord(char)
print(hex(unicode_code))
````

---
#### **3️⃣ UTF-8 (Unicode Transformation Format - 8 bit)**
- **Bit Size**: Variable (1 to 4 bytes per character)
- **Efficiency**:
  - 1 byte for ASCII (0-127)
  - 2 bytes for Latin, Greek, Cyrillic characters
  - 3 bytes for most non-Latin characters
  - 4 bytes for rare symbols and emojis
- **Most widely used encoding on the internet** 🌍.


````python
text = "Hello,世界"
utf8_bytes = text.encode('utf8')
print(utf8_bytes)
print(type(utf8_bytes))
decoded_text = utf8_bytes.decode('utf-8')
print(decoded_text)
````

````python
#-*- coding: utf-8 -*-
print('क्या आप अंग्रेज़ी बोलते हैं?')
````

---

##### 📊 **Comparison Table: ASCII vs Unicode vs UTF-8**

| Feature          | ASCII             | Unicode             | UTF-8                        |
|-----------------|------------------|--------------------|-----------------------------|
| **Bit Size**    | 7-bit             | 8, 16, or 32-bit   | Variable (1-4 bytes)        |
| **Character Set** | English only     | All languages      | All languages               |
| **Storage Size** | Smallest          | Large              | Efficient                   |
| **Backward Compatible** | Yes       | No                 | Yes (for ASCII)             |
| **Best Use Case** | Legacy systems   | Internal processing | Web, files, and APIs        |

---

#### 🔥 **Why is Character Encoding Important in Python?**

##### 1️⃣ Python 3 Uses UTF-8 by Default  
- Python 3 stores all strings as Unicode (UTF-8).
- Encoding errors may occur when working with non-ASCII text.

##### 2️⃣ File Handling & Data Processing
```python
with open("data.txt", "r", encoding="utf-8") as file:
    text = file.read()
print(text)
```

##### 3️⃣ Web Scraping & API Data Handling
```python
import requests
response = requests.get("https://example.com")
print(response.encoding)  # Typically 'utf-8'
print(response.text)
```

##### 4️⃣ Handling Non-ASCII Characters (Multilingual Data)
```python
text = "こんにちは"
encoded_text = text.encode('utf-8')
print(encoded_text)  # UTF-8 bytes

decoded_text = encoded_text.decode('utf-8')
print(decoded_text)  # Back to string
```

##### 5️⃣ Interfacing with Databases & Legacy Systems  
- Databases use different encodings (`utf-8`, `latin-1`, etc.).
- Legacy systems may not support Unicode.

---

#### 🚀 **When Should You Focus on Character Encoding?**
- **🔰 Beginner:** Not crucial unless handling text files.
- **🛠 Intermediate:** Needed for file I/O, APIs, and web scraping.
- **💡 Advanced:** Essential for multilingual apps, NLP, and databases.

---

#### 🎯 **Conclusion**
- ✅ **ASCII**: Simple but limited to English.
- ✅ **Unicode**: Covers all characters but uses more memory.
- ✅ **UTF-8**: Efficient, widely used, and default in Python 3.

💡 **Tip**: Always specify encoding (`utf-8`) when working with files to avoid encoding errors! 🔥

### 🔷 **While**
#### 1️⃣ Print a pattern N times

````python
n = int(input("Enter the number of times you want me to print X"))
my_str = ''
itr = 1
while itr <= n:
    my_str+='X'
    itr+=1
print(f'{my_str}')
````

#### 2️⃣ Finding the square of a number

````python
# finding square of a number  
n = 5 
iter = 0
ans = 0
while iter < n:
    ans += n
    iter+=1
    print(f'iter:{iter},ans:{ans}')
print(f'{n}*{n} = {ans}')
````

### 🔷 **Break**
#### 1️⃣ Find a positive number which is divisible by both 11 and 12

````python
x = 1
while True:
    if x % 11 == 0 and x % 12 == 0:
        break
    x += 1
print(f'The number divisible by both 11 and 12 is : {x}')
````

### 🔷 **For**
```python
for variable in sequence:
    code block
```
⌨️ **Algorithm**

1. The variable following for is bound to the first value in the sequence,and the code block is executed.

2. The variable is then assigned the second value in the sequence, and the code block is executed again.

3. The process continues until the sequence is exhausted or a break statement is executed within the code block. 

````python
total = 0 
for num in (77,11,3):
    total+=num
print(total)
````

3️⃣ Write a program that prints the sum of the prime numbers greater than 2 and less than 1000. 

````python

sum = 0
for n in range(3,1000,2):
    for num in range(2,n):
        if n % num == 0:
            break
        if (n % num != 0) and (num == n-1):
            sum += n      
print(f'The sum of prime numbers is :{sum}')
````
