## <a id='toc1_1_'></a>[🚀 Introduction](#toc0_)

- An “exception” is usually defined as “something that does not conform to the norm,” and is therefore somewhat rare. There is
nothing rare about exceptions in Python.

- Virtually every module in the standard Python library uses them, and Python itself will raise them in many circumstances

- Exception handling allows you to gracefully handle errors in your program.

- An exception does not need to lead to program termination. Exceptions, when raised, can and should be handled by the program

- Sometimes an exception is raised because there is a bug in the program (like accessing a variable that doesn't exist), but many times, an exception is something the programmer can and should anticipate.

- Python provides a convenient mechanism, try-except, for catching and handling exceptions. The general form is
```python
    try:
        code block
    except (list of exception names):
        code block
    else:
        code block
```

```python
    success_failure_ratio = num_successes/num_failures
    print('The success/failure ratio is', success_failure_ratio)
```
- This will fail if there is no failurs , i.e `num_failures = 0`. Attempt to divide by zero will cause the python runtime to raise a `ZeroDivisionError` exception 
```python
try:
    success_failure_ratio = num_successes/num_failures
    print('The success/failure ratio is',success_failure_ratio)
except ZeroDivisionError:
    print('No failures, so the success/failure ratio is undefined.')
```
🔎 Upon entering the `try` block, the interpreter attempts to evaluate the expression `num_successes/num_failures`. If expression evaluation is successful, the program assigns the value of the expression to the variable `success_failure_ratio`, executes the `print` statement at the end of the `try` block, and then proceeds to execute whatever code follows the `try-except` block. If, however, a `ZeroDivisionError` exception is raised during the expression evaluation, control immediately jumps to the `except` block (skipping the assignment and the `print` statement in the `try` block), the `print` statement in the `except` block is executed, and then execution continues following the `try-except` block.

- If it is possible for a block of program code to raise more than one kind of exception, the reserved word except can be followed by a
tuple of exceptions, in which case the except block will be entered if any of the listed exceptions is raised within the try block.
```python
except (ValueError, TypeError):
```
- we can write a separate except block for each kind of exception, which allows the program to choose an action based upon which exception was raised



````python
try:
    print(get_ratios([1, 2, 7, 6], [1, 2, 0, 3]))
    print(get_ratios([], []))
    print(get_ratios([1, 2], ['hello']))
except ValueError as msg:
    print(msg)
    
````

### <a id='toc1_2_3_'></a>[🔢 Logical Errors](#toc0_)

- Logical errors occur when the program does not produce the expected output.
- The code is syntactically correct and does not raise any exceptions.
- Logical errors are the most challenging to debug.

````python
# Logical error: incorrect calculation
x = 10
y = 20
result = x + y  # Should be x * y
````

````python
num1 = input('Enter a number:')
num2 = input('Enter another number:')
sum = num1 + num2
print('The sum of', num1, 'and', num2, 'is', sum)
````

````python
try:
    x = 10 / 0  # Division by zero
except ZeroDivisionError as e:
    print("Error:", e)
````

- In this example, the code inside the `try` block raises a `ZeroDivisionError`.
- The `except` block catches the exception and prints an error message.

````python
import math

a_number = int(input("Please enter an integer "))
# Simulate what happens with negative input
try:
    print(math.sqrt(a_number))
except:
    print("Bad Value for square root")
    print("Using absolute value instead")
    print(math.sqrt(abs(a_number)))
````

### <a id='toc1_2_4_'></a>[🔢 Multiple Exceptions](#toc0_)

````python
try:
    x = 10 / 0  # Division by zero
except ZeroDivisionError as e:
    print("Error:", e)
except Exception as e:
    print("An error occurred:", e)
````

- You can catch multiple exceptions by adding multiple `except` blocks.
- The `except` block that matches the exception type is executed.

### <a id='toc1_2_5_'></a>[🌗 Finally](#toc0_)

````python
try:
    x = 10 / 0  # Division by zero
except ZeroDivisionError as e:
    print("Error:", e)
finally:
    print("Finally block")
````

- The `finally` block is executed regardless of whether an exception occurs.
- The `finally` block is useful for releasing resources or cleaning up.

````python
def read_file(file_path):
    try:
        file = open(file_path, 'r')
        content = file.read()
    except FileNotFoundError as e:
        print(f"Error: {e}")
    else:
        print(f"File content:\n{content}")
    finally:
        # this is to handle the case when the file variable is not initialized
        try:
            file.close()
        except NameError:
            print("File is not open")
        print("Execution of the try-except block is complete.")
````

### <a id='toc1_2_6_'></a>[⛓️ Raise](#toc0_)

- In Python, it is more usual to have a function raise an exception when it cannot produce a result that is consistent with the function's specification

- The Python `raise` statement forces a specified exception to occur

    ```python 
    raise exceptionName(arguments) 
    ```
- `exceptionName` is usually one of the built in exception but programmers can define new exception by creating the subclass of built-in class `Exception`

````python
def get_ratios(vect1, vect2):
    ratios = []
    for index in range(len(vect1)):
        try:
            ratios.append(vect1[index]/vect2[index])
        except ZeroDivisionError:
            ratios.append(float('nan'))
        except:
            raise ValueError
    return
````

````python
def find_an_even(L):
    """
    Assumes L is a list of integers
    Returns the first even number in L
    Raises ValueError if L does not contain an even number
    """
    for num in L:
        if num % 2 == 0:
            return num
    raise ValueError("No even number found in the list.")

find_an_even([1,3,5])
# try: 
#     find_an_even([1,3,5])
# except ValueError as msg:
#     print(msg)
````

````python
def get_grades(fname):
    grades = []
    try:
        with open(fname,'r') as grades_file:
            for line in grades_file:
                try: 
                    grades.append(float(line))
                except:
                    raise ValueError('Cannot convert line to float')
    except IOError:
        raise ValueError(f'get_grades could not open {fname}')
    return grades 

try :
    grades = get_grades('quiz1grades.txt')
    grades.sort()
    median = grades[len(grades)//2]
    print(f'Median grade is {median}')
except ValueError as error_msg:
    print(error_msg)
````

### <a id='toc1_2_7_'></a>[🚙 Common Exception Types](#toc0_)

- **BaseException**: The base class for all built-in exceptions.
- **Exception**: The base class for all non-exit exceptions.
- **ArithmeticError**: The base class for all errors that occur for numeric calculations.
  - **OverflowError**: Raised when the result of an arithmetic operation is too large to be expressed.
  - **ZeroDivisionError**: Raised when division or modulo by zero takes place.
  - **FloatingPointError**: Raised when a floating point operation fails.
- **AttributeError**: Raised when an attribute reference or assignment fails.
- **EOFError**: Raised when the `input()` function hits an end-of-file condition.
- **OSError**: Raised when a system function returns a system-related error.
  - **FileNotFoundError**: Raised when a file or directory is requested but doesn't exist.
  - **PermissionError**: Raised when trying to run an operation without the adequate access rights.
  - **TimeoutError**: Raised when a system function timed out at the system level.
- **RuntimeError**: Raised when an error is detected that doesn't fall in any of the other categories.
  - **NotImplementedError**: Raised when an abstract method that needs to be implemented in an inherited class is not actually implemented.
- **SyntaxError**: Raised when the parser encounters a syntax error.
  - **IndentationError**: Raised when there is an incorrect indentation.
    - **TabError**: Raised when indentation consists of inconsistent tabs and spaces.
**TypeError**: Raised when an operation or function is applied to an object of inappropriate type.
- **ValueError**: Raised when an operation or function receives an argument that has the right type but an inappropriate value.
  - **UnicodeError**: Raised when a Unicode-related encoding or decoding error occurs.
    - **UnicodeEncodeError**: Raised when a Unicode-related error occurs during encoding.
    - **UnicodeDecodeError**: Raised when a Unicode-related error occurs during decoding.
    - **UnicodeTranslateError**: Raised when a Unicode-related error occurs during translation.
- **StopIteration**: Raised by the `next()` function to indicate that there are no further items produced by the iterator.
- **AssertionError**: Raised when an `assert` statement fails.

## <a id='toc1_3_'></a>[📘 Assertions](#toc0_)



- The Python assert statement provides programmers with a simple way to confirm that the state of a computation is as expected

- Assert statement can take 
    
    1. `assert Boolean expression`
    2. `assert Boolean expression, argument`

## <a id='toc1_4_'></a>[🛡️ Introduction](#toc0_)

When an `assert` statement is encountered, the Boolean expression is evaluated.  
If it evaluates to `True`, execution proceeds on its merry way.  
If it evaluates to `False`, an `AssertionError` exception is raised.

**Assertions** are a useful defensive programming tool.  
They can be used to confirm that the arguments to a function are of appropriate types.  
They are also a useful debugging tool. They can be used, for example, to confirm that intermediate values have the expected values or that a function returns an acceptable value.

---

````python
x = 1
assert x > 0, "x must be positive"
print("All good!")
````

````python
x = -5
assert x > 0, "x must be positive"
# Raises: AssertionError: x must be positive
````

````python
def divide(a, b):
    assert isinstance(a, (int, float)), "a must be a number"
    assert isinstance(b, (int, float)), "b must be a number"
    assert b != 0, "b must not be zero"
    return a / b

print(divide(10, 2))  
print(divide(10, 0))  
````

⚠️ When not to use assert

Assertions are not a replacement for proper error handling in production code.
They can be disabled globally with the -O (optimize) flag when running Python, so do not use them for critical validation (e.g., checking user input).

___

✅ Use assert for:
- Catching bugs early in development

- Documenting assumptions in code

- Quick sanity checks during debugging
___

````python
def get_temperature():
    # Simulate reading from a sensor
    return -273

temp = get_temperature()
assert temp > -273.15, "Temperature below absolute zero!"
````

🚨 Why Not Use assert for Production Error Handling?
1. Assertions can be disabled.
    - Python’s assert statements are stripped out when code is run with the -O (optimize) flag:

    `python -O script.py`
2. They don’t raise specific exceptions.
    - assert always raises a general AssertionError, which gives little context or control compared to custom exceptions or handling with try-except.
3. They’re not user-friendly.
    - assert crashes the program immediately. In production, you want to handle errors gracefully, not abruptly.

⚖️ Comparing assert vs Proper Error Handling

❌ Using assert improperly for production validation

````python
def process_payment(amount):
    assert amount > 0, "Amount must be greater than zero"
    # proceed with processing
    print(f"Processing payment of ₹{amount}")
process_payment(-50)
````

✅ Proper error handling using exceptions

````python
def process_payment(amount):
    if amount <= 0:
        raise ValueError("Amount must be greater than zero")
    print(f"Processing payment of ₹{amount}")
````
