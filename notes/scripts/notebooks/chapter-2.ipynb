## 🌀 Exhaustive Enumeration

### 🧪 Find the cube root


```python
import time 
x = int(input("Enter an integer: "))
start = time.time()
ans = 0
while ans**3 < abs(x):
    ans += 1
if ans**3 != abs(x):
    print(f'{x} is not a perfect cube')
else:
    if x < 0:
        ans = -ans
    print(f'{ans} is the cuberoot of {x}')
end = time.time()
print(f'total time {(end-start):0.4f}')
```

* The algorithmic technique used in this program is a variant of guess-and-check called exhaustive enumeration

* We enumerate all possibilities until we get to the right answer or exhaust the space of possibilities

* Exhaustive enumeration algorithms are often the most practical way to solve a problem

### 🧪 Method to find prime number

The simplest way to find out if an integer, x, greater than 3 is prime, is to divide x by each integer between 2 and, x-1. If the
remainder of any of those divisions is 0, x is not prime, otherwise x is prime

```python
num = int(input("Enter the integer greater than 2:"))
smallest_divisor = None
for guess in range(2,num):
    if num % guess == 0:
        smallest_divisor = guess
        break
if smallest_divisor is None:
    print(f'The number {num} is a prime number')
else:
    print(f'The smallest divisor is {smallest_divisor}')
```

### 📚 Searching for a Word in a Hard-Copy Dictionary

#### 🚀 Introduction
Consider the problem of discovering whether a word starting with a given sequence of letters appears in a hard-copy dictionary of the English language. 

One possible approach is **exhaustive enumeration**, where we start at the first word and examine each word sequentially until we either find a match or reach the end of the dictionary.

#### 🌀 Exhaustive Enumeration and Search Complexity
- If the dictionary contains **$ n$ words**, we perform a **linear search**.
- If the word exists in the dictionary, we will find it at some position $ k $ where $ k $ is uniformly distributed between $ 1 $ and $ n $.
- On average, the search will take ** $ n/2 $ probes** before finding the word.
- If the word is **not present**, we will have to check all ** $n$ words**, leading to a worst-case of  $n $ probes.

#### 🏹 Mathematical Justification of $ n/2 $
Let $X $ be the position at which we find the word. Since the word can be at any position **uniformly** at random:

$$
P(X = k) = \frac{1}{n}, \quad \text{for } k = 1, 2, ..., n.
$$

The expected number of probes required to find the word is:

$$
E[X] = \sum_{k=1}^{n} k \cdot P(X = k)
$$

Substituting $ P(X = k) = \frac{1}{n} $:

$$
E[X] = \frac{1}{n} \sum_{k=1}^{n} k
$$

Using the formula for the sum of the first \( n \) natural numbers:

$$
\sum_{k=1}^{n} k = \frac{n(n+1)}{2}
$$

we get:

$$
E[X] = \frac{1}{n} \times \frac{n(n+1)}{2} = \frac{n+1}{2}
$$

For large \( n \), this simplifies to:

$$
E[X] \approx \frac{n}{2}
$$

Thus, on average, we need to examine **$ n/2 $ words** before finding the desired word.



#### 🔨 Worst-Case Analysis
If the word is **not present in the dictionary**, we must check **all $ n $ words** before concluding that it does not exist.

Thus, the worst-case scenario requires **$ n $ probes**.

#### 🎯 Real-Life Intuition
Those who have had the pleasure of looking up a word in a **physical dictionary** rather than an online one would never proceed in this exhaustive manner. Instead, we typically:
- Use **alphabetical ordering** to jump to an approximate location.
- Use **binary search**-like techniques by flipping to a section and refining our search.
- This approach significantly reduces the number of probes needed.

# 📘 Bisection Search

* An example of sucessive approximation method


```python
x = int(input("Enter an integer: "))
epsilon = 0.01
num_guesses = 1
low = 0
high = max(1.0, x)
ans = (high + low) / 2.0
while abs(ans**2 - x) >= epsilon:
    print(f'low:{low}, high:{high}, ans:{ans}, guess:{num_guesses}')
    if ans**2 > x:
        high = ans 
    else:
        low = ans
    ans = (high+low)/2
    num_guesses += 1
```


## **Introduction**
The term **"bisection"** comes from the **Latin word "bi-" (meaning "two")** and **"section" (meaning "cut")**. 

In **bisection search**, the search space is **cut in half (divided into two parts) at each step**, which is why the name **bisection** is used.

### **1. Start with an Interval**
- Suppose we want to find a number \( y \) such that $$ f(y) = 0 $$ in an interval $$ [a, b] $$.

### **2. Find the Midpoint**
- Compute the midpoint:

  $$
  m = \frac{a + b}{2}
  $$

### **3. Decide Which Half to Keep**
- If $$ f(m) $$ is close enough to zero, we stop.
- If $$ f(m) $$ has the **same sign** as $$ f(a) $$, then the root must be in $$ [m, b] $$, so we **cut away the left half** and update $$ a = m $$.
- Otherwise, the root must be in $$ [a, m] $$, so we **cut away the right half** and update $$ b = m $$.

### **4. Repeat Until Convergence**
- Each step **bisects the interval**, reducing its size by half.
- The search space shrinks **exponentially**, ensuring convergence.

# **Proof: Bisection Search Shrinks the Search Space Exponentially**

## **1. Defining the Problem**
Suppose we are searching for a value (e.g., a root of a function) in an **initial search interval**:

$$
[a_0, b_0]
$$

where:

$$
a_0 \text{ is the lower bound,}
$$

$$
b_0 \text{ is the upper bound,}
$$

$$
L_0 = b_0 - a_0 \text{ is the initial interval length.}
$$

At each step, we compute the **midpoint**:

$$
m_k = \frac{a_k + b_k}{2}
$$

and decide which half to keep based on the function's sign.

---

## **2. How the Interval Shrinks**
In **each iteration**, the new interval **halves**:
- If the root is in the left half, update:

  $$
  b_{k+1} = m_k
  $$

- If the root is in the right half, update:

  $$
  a_{k+1} = m_k
  $$

The **length of the interval** after $ k $ iterations is:

$$
L_k = \frac{L_0}{2^k}
$$

where $ L_0$ is the initial length of the interval.

This shows that **the interval length decreases exponentially** with base 2.

---

## **3. Exponential Decay**
From the formula:

$$
L_k = \frac{L_0}{2^k}
$$

- The search space shrinks **exponentially** because:

  $$
  2^k \text{ grows exponentially.}
  $$

- After $k$ iterations, the remaining search space is reduced to:

  $$
  \frac{1}{2^k} \text{ of the original space.}
  $$

---

## **4. Number of Iterations Needed for Convergence**
Suppose we want to stop when the interval size is smaller than a small error tolerance $ \epsilon$, meaning:

$$
L_k < \epsilon
$$

Using the shrinking formula:

$$
\frac{L_0}{2^k} < \epsilon
$$

Solving for $ k $:

$$
k > \log_2 \left(\frac{L_0}{\epsilon}\right)
$$

Thus, the number of iterations required to achieve accuracy $ \epsilon$ is:

$$
k = \lceil \log_2 (L_0 / \epsilon) \rceil
$$

This confirms **exponential convergence**, as the number of iterations is proportional to the logarithm of the precision.

---

## **Conclusion**
- **The search space shrinks exponentially** as:

  $$
  L_k = \frac{L_0}{2^k}
  $$

- **The number of iterations needed for accuracy**  $\epsilon$ **is logarithmic** in the initial interval size.
- **Bisection search is very efficient**, requiring at most:

  $$
  O(\log_2 (L_0 / \epsilon))
  $$

  iterations.

🚀 This is why **bisection search converges quickly** compared to linear search methods!

### 1. Decimal vs. Binary Representation

* **Decimal (Base 10):** The number system we commonly use.
* **Binary (Base 2):** The number system used by computers.
* **Key Issue:** Not all decimal fractions can be represented exactly in binary.

**Example:**

* 0.1 (decimal) = 0.0001100110011... (repeating binary)

---

### 2. Converting Decimal to Binary

**Steps:**

1.  **Separate Integer and Fractional Parts:** If applicable.
2.  **Integer Part:** Repeatedly divide by 2, recording remainders. Read remainders in reverse order.
3.  **Fractional Part:** Repeatedly multiply by 2, recording integer parts. Read integer parts in forward order.
4.  **Combine:** Integer.Fractional

**Example:**

* 0.5 (decimal) = 0.1 (binary)
* 0.6 (decimal) = 0.100110011... (repeating binary)

**Example 1: Converting 3.75 to Binary**

* **Integer Part (3):**
    * 3 / 2 = 1 (remainder 1)
    * 1 / 2 = 0 (remainder 1)
    * Binary: 11
* **Fractional Part (0.75):**
    * 0.75 * 2 = 1.5 (1)
    * 0.5 * 2 = 1.0 (1)
    * Binary: 11
* **Combined:** 3.75 (decimal) = 11.11 (binary)

**Example 2: Converting 0.6 to Binary**

* **Fractional Part (0.6):**
    * 0.6 * 2 = 1.2 (1)
    * 0.2 * 2 = 0.4 (0)
    * 0.4 * 2 = 0.8 (1)
    * 0.8 * 2 = 1.6 (1)
    * 0.6 * 2 = 1.2 (1)
    * 0.2 * 2 = 0.4 (0)
    * ... (repeating pattern)
    * Binary: 0.100110011... (repeating)
* **Combined:** 0.6 (decimal) = 0.100110011... (binary)

---

### 3. Floating-Point Representation

* Computers use a finite number of bits to store floating-point numbers.
* This leads to approximations and rounding errors.

---

### 4. IEEE 754 Standard

* A standard for representing floating-point numbers in computers.
* Defines formats for single-precision (32-bit) and double-precision (64-bit).
* Structure: Sign bit, Exponent, Significand.

**Key Features:**

* **Sign:** 0 for positive, 1 for negative.
* **Exponent:** Biased to simplify comparisons.
* **Significand:** Significant digits of the number.

**Example: Single-Precision**

* 1 bit (sign) + 8 bits (exponent) + 23 bits (significand) = 32 bits


##### IEEE 754 Example: Representing 5.625 in Single-Precision

Let's break down how the decimal number 5.625 is represented in IEEE 754 single-precision (32-bit).

**1. Convert to Binary:**

* **Integer Part (5):**
    * 5 / 2 = 2 (remainder 1)
    * 2 / 2 = 1 (remainder 0)
    * 1 / 2 = 0 (remainder 1)
    * Binary: 101
* **Fractional Part (0.625):**
    * 0.625 * 2 = 1.25 (1)
    * 0.25 * 2 = 0.5 (0)
    * 0.5 * 2 = 1.0 (1)
    * Binary: 101
* **Combined:** 5.625 (decimal) = 101.101 (binary)

**2. Normalize:**

* Move the binary point to the left until there's only one '1' to the left of the point.
* 101.101 = 1.01101 x 2^2

**3. Determine Components:**

* **Sign:** 0 (positive number)
* **Exponent:** 2. However, we need to add the bias (127 for single-precision): 2 + 127 = 129. Convert 129 to binary: 10000001.
* **Significand:** Remove the leading '1' from the normalized binary and pad with zeros to fill the 23 bits: 01101000000000000000000.

**4. Combine Components:**

* **Sign (1 bit):** 0
* **Exponent (8 bits):** 10000001
* **Significand (23 bits):** 01101000000000000000000
* **Final IEEE 754 Representation:** 0 10000001 01101000000000000000000

### 5. Exponent Bias

* **Purpose:** Simplifies floating-point comparisons.
* **How it Works:** Shifts the exponent range to make it positive.
* **Single-Precision Bias:** 127.
* **Calculation:** 2^(exponent_bits - 1) - 1.

**Why 127?**

* 8 bits for the exponent allow 256 values.
* Bias centers the exponent range around zero.
* 127 = 2^(8-1) -1.

**Benefits:**

* Allows treating floating-point numbers as integers for comparisons.
* Handles negative exponents efficiently.
* Ensures consistent ordering.

**Example:**

* Exponent 0 is represented as 127 in biased form.
* Exponent -1 is represented as 126 in biased form.

---

### 6. Consequences and Best Practices

* **Rounding Errors:** Be aware of them.
* **Equality Tests:** Avoid direct `==` comparisons. Use tolerance-based comparisons (e.g., `abs(x - y) < tolerance`).
* **Accumulation:** Rounding errors can accumulate in calculations.
* **Use `round()`:** To control decimal place precision when printing.
* **Understand limitations:** No finite binary representation can exactly represent all decimal fractions.

# Newton-Raphson Method and Roots of Polynomials

## Introduction

This section explores the Newton-Raphson method for finding roots of polynomials and discusses the significance of these roots.

## Newton-Raphson Method

The Newton-Raphson method is an iterative technique for finding successively better approximations to the roots (or zeroes) of a real-valued function.

### Key Concepts

* **Polynomial:** An expression consisting of variables and coefficients, involving only the operations of addition, subtraction, multiplication, and non-negative integer exponents of variables. Example: $3x^2 + 2x + 3$.
* **Root:** A root of a polynomial $p(x)$ is a value 'r' such that $p(r) = 0$.
* **Derivative:** The derivative of a function $p(x)$, denoted as $p'(x)$, represents the rate of change of the function. For a term $cx^p$, the derivative is $cpx^{p-1}$.
* **Iteration:** The method works by repeatedly refining an initial guess until it gets sufficiently close to the actual root.

### The Formula

The core of the Newton-Raphson method is this iterative formula:

$guess_{new} = guess_{old} - \frac{p(guess_{old})}{p'(guess_{old})}$

Where:

* $guess_{old}$ is the current approximation of the root.
* $guess_{new}$ is the improved approximation.
* $p(guess_{old})$ is the value of the polynomial at $guess_{old}$.
* $p'(guess_{old})$ is the value of the derivative of the polynomial at $guess_{old}$.

### Applying it to Square Root Calculation

Finding the square root of a number 'k' can be expressed as finding the root of the polynomial $p(x) = x^2 - k$.

1.  **Polynomial and Derivative:**
    * $p(x) = x^2 - k$
    * $p'(x) = 2x$

2.  **Newton-Raphson Formula:**
    * $guess_{new} = guess_{old} - \frac{guess_{old}^2 - k}{2 * guess_{old}}$

# Proof of Newton-Raphson

The proof is rooted in calculus and the idea of linear approximation.

## Geometric Intuition

Imagine you have a function $f(x)$ and you want to find its root (where $f(x) = 0$).

* **Tangent Line:** At $x_n$, you find the tangent line to the graph of $f(x)$.
* **X-intercept:** You find where this tangent line intersects the x-axis. This x-intercept is your next guess, $x_{n+1}$.
* **Iteration:** You repeat this process.

## Mathematical Derivation

### Tangent Line Equation:

The equation of the tangent line to $f(x)$ at $x_n$ is given by:

$y - f(x_n) = f'(x_n)(x - x_n)$

### Finding the X-intercept:

To find the x-intercept, we set $y = 0$:

$0 - f(x_n) = f'(x_n)(x_{n+1} - x_n)$

### Solving for x:

Rearrange the equation to solve for $x_{n+1}$:

$-f(x_n) = f'(x_n)(x_{n+1} - x_n)$

$\frac{-f(x_n)}{f'(x_n)} = x_{n+1} - x_n$

$x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}$

## Taylor Series Perspective

### Taylor Series Expansion:

The Taylor series expansion of $f(x)$ around $x_n$ is:

$f(x) \approx f(x_n) + f'(x_n)(x - x_n) + \frac{f''(x_n)}{2!}(x - x_n)^2 + ...$

### Linear Approximation:

If $x$ is close to $x_n$, we can ignore the higher-order terms and approximate $f(x)$ as:

$f(x) \approx f(x_n) + f'(x_n)(x - x_n)$

### Finding the Root:

We want to find $x$ such that $f(x) = 0$, so:

$0 \approx f(x_n) + f'(x_n)(x_{n+1} - x_n)$

### Solving for $x_{n+1}$:

$x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}$

## Conditions for Convergence

* $f(x)$ must be continuous and differentiable in the interval of interest.
* $f'(x_n)$ should not be zero near the root.
* The initial guess $x_0$ should be sufficiently close to the root.
