## <a id='toc1_1_'></a>[🚀 Introduction](#toc0_)

- The key to object-oriented programming is thinking about objects as **collections of both data and the methods that operate on that data**.

## <a id='toc1_2_'></a>[🌀 Abstract Data Types and Classes](#toc0_)

- An abstract data type is a set of objects and the operations on those objects
- These are bound together so that programmers can pass an object from one part of a program to another, and in doing so provide access not only to the data attributes of the object but also to operations that make it easy to manipulate that data

- The specifications of those operations define an interface between the abstract data type and the rest of the program. The interface defines the behavior of the operations—what they do, but not how they do it. The interface thus provides an abstraction barrier that isolates the rest of the program from the data structures, algorithms, and code involved in providing a realization of the type **abstraction**.

- In Python, we implement the data abstractions using **classes**

- Each class definition begins with the reserved word `class` followed by the name of the class and some information about how it relates to other classes


```python
class Toy(object):
    def __init__(self):
        self._elems = []
    def add(self, new_elems):
        self._elems += new_elems
    def size(self):
        return len(self._elems)

print(type(Toy))
print(type(Toy.__init__), type(Toy.add), type(Toy.size))
```

- Python has a number of special function names that start and end with two underscores. These are often referred to as **magic methods**

- Whenever a class is instantiated, a call is made to the `__init__` function defined in that class

- When the line of code `s = Toy()` is executed, the interpreter will create a new instance of type `Toy`,and then call `Toy.__init__` with the newly created object as the actual parameter that is bound to the formal parameter `self`. When invoked, `Toy.__init__` creates the list object `_elems`, which becomes part of the newly created instance of type `Toy`.

- The list`_elems` is called a **data attribute** of the instance of `Toy`

```python
t1 = Toy()
print(type(t1))
print(type(t1.add))
t2 = Toy()
print(t1 is t2)
```

`t1.add` is of type method, whereas `Toy.add` is of type function. Because `t1.add` is a method, we can invoke it

- class attributes are deined in a calss function

- When the class is instantiated, e.g., by the statement `t = Toy()`, instance attributes, e.g., `t.size`, are created

- While `t.size` is initially bound to the size function defined in the class `Toy`, that binding can be changed during the course of a
computation. For example, you could (but definitely should not!) change the binding by executing `t.size = 3`.

- When data attributes are associated with a class, we call them **class variables**. When they are associated with an instance, we
call them **instance variables**. 
    - For example,` _elems` is an instance variable because for each instance of class `Toy`, `_elems` is bound to a different list


```python
t1 = Toy()
t2 = Toy()
t1.add([3, 4])
t2.add([4])
print(f'{t1.size()}, {t2.size()}')
```

- `Toy().add` method has two formal parameters, but we appear to be calling it with only one actual parameter
- This is an artifact of using dot notation to invoke a method associated with an instance of a class 
- The object associated with the expression before the dot is implicitly passed as the first parameter to the method 
- `self` as the name of the formal parameter to which this actual parameter is bound
-  common convention is to start the name of data attributes with an underscore.
-  leading `_` to indicate that the attribute is private to the class, i.e., should not be directly accessed outside the class
    - Python doesn’t enforce access restrictions like private or protected in Java/C++, so this is a naming convention

```python
class Int_set(object):
    
    def __init__(self):
        self._vals=[]
    def insert(self, e):
        if e not in self._vals:
            self._vals.append(e)
    def member(self,e):
        return e in self._vals
    def remove(self, e):
        try:
            self._vals.remove(e)
        except ValueError:
            print(f'ValueError:value not found')
    def get_members(self):
        return self._vals[:]
    def __str__(self):
        if self._vals == []: 
            return '{}'
        self._vals.sort()
        result = ''
        for e in self._vals:
            result += f'{str(e)},'
        return '{result[:-1]}'
        
        
```

```python
s = Int_set()
s.insert(2)
print(s.member(2))
```

🧠 What is Data Abstraction?
- Data abstraction means hiding the internal details of how data is stored or implemented and exposing only the necessary operations (methods) to interact with it.

🔍 What is Representation-Independence?
- Representation-independence means the users of a class don’t need to know how something is stored internally—only what the methods do. This gives you freedom to change the internal implementation later without changing the way users interact with the object
- The set is internally represented by a Python list named `_vals`.

- This is a private representation (the _ indicates it shouldn't be accessed directly).

- You could change this to a dictionary, set, or anything else later—as long as your methods behave the same way to the outside world.

### <a id='toc1_3_1_'></a>[🎯 Python's Design Goal](#toc0_)

Python aims to make **user-defined types** feel as natural and easy to use as **built-in types**.  
This is achieved through **magic methods** (also called **dunder methods**, because they have double underscores).

---

### <a id='toc1_3_2_'></a>[✨ What Are Magic Methods?](#toc0_)

Magic methods let us define custom behavior for:
- Built-in functions like `str()` or `len()`
- Operators like `+`, `==`, `<`, etc.

They make your classes work more like built-in Python objects.

### <a id='toc1_3_3_'></a>[🔍 Why Use Magic Methods?](#toc0_)

They let your class:
- Work like a native Python type
- Be printable, comparable, addable, iterable, etc.
- Be used in collections like set, dict, etc.

---

### <a id='toc1_3_5_'></a>[➕ Magic Methods for Operators](#toc0_)

You can override Python’s **infix operators** to define custom behavior in your classes using magic methods.

#### <a id='toc1_3_5_1_'></a>[🔁 Operator to Magic Method Mapping](#toc0_)

| Operator | Magic Method     |
|----------|------------------|
| `+`      | `__add__`        |
| `-`      | `__sub__`        |
| `*`      | `__mul__`        |
| `/`      | `__truediv__`    |
| `//`     | `__floordiv__`   |
| `%`      | `__mod__`        |
| `**`     | `__pow__`        |
| `&`      | `__and__`        |
| `^`      | `__xor__`        |
| `<<`     | `__lshift__`     |
| `>>`     | `__rshift__`     |
| `==`     | `__eq__`         |
| `!=`     | `__ne__`         |
| `<`      | `__lt__`         |
| `<=`     | `__le__`         |
| `>`      | `__gt__`         |
| `>=`     | `__ge__`         |


```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    def __str__(self):
        return f"<{self.x}, {self.y}>"

v1 = Vector(1, 2)
v2 = Vector(3, 4)
print(v1 + v2)  # Output: <4, 6>
```

but don't 😅

> Stick to the conventional meaning of operators to keep your code understandable and maintainable.

#### <a id='toc1_3_5_3_'></a>[🔐 Hashable Types](#toc0_)
To make your object usable as:

- A key in a dict
- An element in a set

It must:

1. Implement `__hash__()`

2. Be immutable

3. Implement `__eq__()` consistently with `__hash__()`

```python
class Point:
    def __init__(self, x, y):
        self._x = x
        self._y = y
    
    def __eq__(self, other):
        return self._x == other._x and self._y == other._y
    
    def __hash__(self):
        return hash((self._x, self._y))

# Can be used as dictionary key
d = {Point(1, 2): "A"}
```

```python
class Toy(object):
    def __init__(self):
        self._elems = []
    def add(self,new_elems):
        self._elems += new_elems
    def __len__(self):
        return len(self._elems)
    def __add__(self,other):
        new_toy = Toy()
        new_toy._elems = self._elems + other._elems
        return new_toy
    def __eq__(self, other):
        return self._elems == other._elems
    def __str__(self):
        return str(self._elems)
    def __hash__(self):
        return id(self)

t1 = Toy()
t2 = Toy()
t1.add([1,2])
t2.add([3,4])
t3 = t1 + t2 
print(f'The value of t3 is {t3}')
print(f'the length of t3 is {len(t3)}')
d = {t1:'A',t2:'B'}
print('The value', d[t1], 'is associated with the key t1 in d')
```

#### <a id='toc1_3_5_4_'></a>[❌ Is Not Advisable](#toc0_)

```python
toy1 = Toy()
toy2 = Toy()
toy1.add([1,2])
toy2.add([1,2])
print(toy1 == toy2)
print(toy2.__hash__() == toy1.__hash__())
```

- Even though their contents are the same, their `hash()` values are not equal which is an important condition for using this as keys in **dictionary** or values in **set**

#### <a id='toc1_3_5_5_'></a>[🧪 Person Class Example](#toc0_)

```python
import datetime
class Person(object):
    def __init__(self, name):
        self._name = name 
        try:
            last_blank = name.rindex(' ')
            self._last_name = name[last_blank+1:]
        except:
            self._last_name = name 
        self.birthday = None 
    
    def get_name(self):
        return self._name
    def get_last_name(self):
        return self._last_name
    def set_birthday(self,birthdate):
        self.birthday = birthdate
    def get_age(self):
        if self.birthday == None:
            raise ValueError
        age = datetime.date.today() - self.birthday
        return age 
    def __lt__(self,other):

        if self._last_name == other._last_name:
            return self._name < other._name 
        return self._last_name < other._last_name
    def __str__(self):
        return self._name
```

```python
me = Person('Michael Guttag')
him = Person('Barack Hussein Obama')
her = Person('Madonna')
print(him.get_last_name())
him.set_birthday(datetime.date(1961, 8, 4))
her.set_birthday(datetime.date(1958, 8, 16))
print(him.get_name(), 'is', him.get_age(), 'days old')
```

## <a id='toc1_4_'></a>[👪 Inheritence](#toc0_)

- Inheritance provides a convenient mechanism for building groups of related abstractions
- It allows programmers to create a type hierarchy in which each type inherits attributes from the types above it in the hierarchy

```python
class MIT_person(Person):
    _next_id_num = 0
    def __init__(self, name):
        super().__init__(name)
        self._id_num = MIT_person._next_id_num
        MIT_person._next_id_num +=1
    def get_id_num(self):
        return self._id_num
    def __lt__(self, other):
        return self._id_num < other._id_num
    # def is_student(self):
    #     return isinstance(self, Student)
    
```

**🤔What is happening** 

1. The class `MIT_person` inherits attributes from its parent class, `Person`, including all of the attributes that `Person` inherited from its parent class, `object`

2.  In the jargon of object-oriented programming, `MIT_person` is a subclass of `Person`, and therefore inherits the attributes of its superclass

3. Subclass can add new attributes : For example, the subclass `MIT_person` has added the class variable `_next_id_num`, the instance variable
`_id_num`, and the method `get_id_num`

4. Override- replace, attributes of the superclass. For example, `MIT_person` has overridden `__init__` and `__lt__`. 
    1. When a method has been overridden, the version of the method that is executed is based on the object used to invoke the method. 
    2. If the type of the object is the subclass, the version defined in the subclass will beused. 
    3. If the type of the object is the superclass, the version in the superclass will be used
5. The method `MIT_person.__init__` first uses `super().__init__(name)` to invoke the` __init__` function of its super class (`Person`). This initializes the inherited instance variable `self._name`. It then initializes `self._id_num`, which is an instance variable that instances of `MIT_person` have but instances of `Person` do not.

6. The instance variable `self._id_num` is initialized using a class variable, `_next_id_num`, that belongs to the class `MIT_person`, rather than to instances of the class. When an instance of `MIT_person` is
created, a new instance of `next_id_num` is not created. This allows `__init__` to ensure that each instance of MIT_person has a unique `_id_num`.
---


```python
p1 = MIT_person('Barbara Beaver')

print(str(p1) + '\'s id number is ' + str(p1.get_id_num()))
```

1. The first line creates a new `MIT_person`. 

2. The second line attempts to evaluate the expression `str(p1)`, the runtime system first checks to see if there is an` __str__` method associated with class `MIT_person`. Since there is not, it next checks to
see if there is an `__str__` method associated with the immediate superclass, `Person`, of `MIT_person`. There is, so it uses that. 

3. When the runtime system attempts to evaluate the expression `p1.get_id_num()`, it first checks to see if there is a `get_id_num` method associated with class `MIT_person`. There is, so it invokes that method and prints

```python
p1 = MIT_person('Mark Guttag')
p2 = MIT_person('Billy Bob Beaver')
p3 = MIT_person('Billy Bob Beaver')
p4 = Person('Billy Bob Beaver')
```

```python
print('p1 < p2 =', p1 < p2)
print('p3 < p2 =', p3 < p2)
print('p4 < p1 =', p4 < p1)
```

1. Since `p1`, `p2`, and `p3` are all of type `MIT_person`, the interpreter will
use the `__lt__` method defined in class `MIT_person` when evaluating
the first two comparisons, so the ordering will be based on
identification numbers. 

2. In the third comparison, the `<` operator is applied to operands of different types. Since the first argument of the expression is used to determine which `__lt__` method to invoke, `p4 < p1` is shorthand for `p4.__lt__(p1)`. Therefore, the interpreter uses the `__lt__` method associated with the type of `p4` `Person`, and the “people” will be ordered by name.

```python
print('p1 < p4 =', p1 < p4)
```

❌ The runtime system will invoke the `__lt__`operator associated with the type of `p1`, i.e., the one defined in class `MIT_person`.

### <a id='toc1_4_2_'></a>[🤾‍♂️ The Liiskov Substitution Principle](#toc0_)

- During overrides, make sure that important behaviors of the supertype must be supported by each of its subtypes

- If client code works correctly using an instance of the supertype, it should also work correctly when an instance of the subtype is substituted (hence the phrase substitution principle) for the instance of the supertype 

- For example, it should be possible to write client code using the specification of `Student` and have it work correctly on a
`TransferStudent`

- Conversely, there is no reason to expect that code written to work for `TransferStudent` should work for arbitrary types of `Student`

```python
class Grades(object):
    def __init__(self):
        self._students = []
        self._grades = {}
        self._is_sorted = True 
    def add_student(self, student:Student):
        """Add student to the grade book

        Args:
            student (_Student_): A Student object 

        Raises:
            ValueError: _description_
        """
        if student in self._students:
            raise ValueError("Duplicate Student")
        self._students.append(student)
        self._grades[student.get_id_num()] = [] 
        self._is_sorted = False
        
    def add_grade(self, student:Student, grade:float):
        """Add the grade in to a student grade list

        Args:
            student (Student): An instance of student class
            grade (float): a floating point value

        Raises:
            ValueError: _description_
        """
        try:
            self._grades[student.get_id_num()].append(grade)
        except:
            raise ValueError("No student exists")
    
    def get_grades(self, student:Student):
        """Return list of grades in the grade book

        Args:
            student (Student ): An instance of the Student class
        """
        try:
            return self._grades[student.get_id_num()][:]
        except:
            raise ValueError("No student found")
            
    def get_students(self):
        """Returns the sorted list of the students in the grade book
        """
        if not self._is_sorted:
            self._students.sort()
            self._is_sorted = True
        return self._students[:]
        
```

```python
course = Grades()
course.add_student(Grad('Bernie'))
course.add_student(Grad('nithin'))
for s in course.get_students():
    print(s)
```

```python
print(course.get_students()[:][0])
```

- If `get_students` returned `self._students`, the last line of code
would have the (probably unexpected) side effect of changing the set
of students in `course`.

```python
def grade_report(course:Grades):
    """return a grade report for a specific course say `six_hundered`

    Args:
        course (Grades): _description_
    """
    report =''
    for s in course.get_students():
        tot = 0.0
        num_grades = 0
        for g in course.get_grades(s):
            tot += g 
            num_grades += 1
        try: 
            average = tot/num_grades
            report += f"{s}'s average grade is {average}\n"
        except ZeroDivisionError:
            report += f"{s} has no grades\n"
    return report
```

### <a id='toc1_4_3_'></a>[🛠️ Encapsulation and Information Hiding](#toc0_)

#### <a id='toc1_4_3_1_'></a>[🏰 What is encapsulation ?](#toc0_)

- Bundling together of data attributes and methods - a `class` for operating on them and restricting direct access to some of the object’s components.

- Encapsulation is like putting your data (variables) and code (functions) inside a box (a class) — and deciding what you want to expose outside the box and what you want to keep hidden.

#### <a id='toc1_4_3_2_'></a>[🕵️‍♀️ What is Information Hiding ?](#toc0_)
- Information Hiding is a software design principle where internal details of a system or component are kept hidden from the outside world, exposing only what is necessary

- Some programming languages (Java and C++, for example)
provide mechanisms for enforcing information hiding. Programmers
can make the attributes of a class **private**, so that clients of the class
can access the data only through the object's methods

-  Python 3 uses
a naming convention to make attributes invisible outside the class.
When the name of an attribute starts with `__ `(**double underscore**)
but does not end with `__`, that attribute is not visible outside the
class.


```python
class info_hiding(object):
    def __init__(self):
        self.visible = "Look at me"
        self.__also_visible__ = "Look at me too"
        self.__invisible = "Don't look at me directly"
    
    def print_visible(self):
        print(self.visible)
    
    def print_invisible(self):
        print(self.__invisible)
    
    def __print_invisible(self):
        print(self.__invisible)
    
    def __print_invisible__(self):
        print(self.__invisible)
    
```

```python
test = info_hiding()
print(test.visible)
print(test.__also_visible__)
print(test.__invisible)
```

```python
test.print_invisible()
test.__print_invisible__()
test.__print_invisible()
```

### <a id='toc1_4_4_'></a>[📝 Important Note](#toc0_)
-  When a subclass attempts to use a hidden attribute of its
superclass, an AttributeError occurs. This can make using
information hiding using `__` a bit cumbersome

- Because it can be cumbersome, many Python programmers do not take advantage of the `__` mechanism for hiding attributes

- Python allows acessing or reading the instance and class variables from outside the  the class definition but also lets program to write these variables

- We avoid directly accessing data attributes because it is
dangerous for client code to rely upon something that is not part of
the specification, and is therefore subject to change


- this relatively weak static semantic checking is a flaw in
Python, it is not a fatal flaw

- A disciplined programmer can simply follow the sensible rule of not directly accessing data attributes from outside the class in which they are defined

### <a id='toc1_4_5_'></a>[🎰 Generators](#toc0_)

-  In function `grade_report`, the invocation of `course.get_students` creates and returns a list of size `n`, where `n` is the number of students. This is probably not a problem for a grade book for a single class, but imagine keeping track of the grades of 1.7 million high school students taking the SATs 

- Creating a new list of that size when the list already exists is a significant inefficiency
    -    One solution is to abandon the abstraction and allow `grade_report` to directly access the instance variable `course.students` but that would violate **information hiding**

- Better Solution : **generator**
    - Any function definition containing a `yield` statement is treated in
a special way. The presence of `yield` tells the Python system that the
function is a generator. Generators are typically used with `for`
statements

🔁 What Are Generators?
A generator is a function that:
- Uses yield to produce a sequence of values one at a time.
- Remembers its state between calls.
- Is lazy — it generates values on demand, not all at once like lists.

💡 Why Use Generators?
- Save memory — no need to store all values in memory.
- Efficient for large datasets or infinite sequences.
- Readable and elegant way to write iterators.

```python

class test_student(object):
    def __init__(self, students):
        self._students = students
    def get_students(self):
        """Return the students int he grade book one at a time in alphabetical order
        """
        self._students.sort()
        for s in self._students:
            yield s
```

```python
students = ['nithin', 'ravi', 'anoop']
st1 = test_student(students)
for s in st1.get_students():
    print(s)
```

🔍 Difference from `return`:
- return exits the function.
- yield pauses the function, saving its state, and resumes from there the next time.

### <a id='toc1_5_1_'></a>[🏠 Understanding Mortgages: A Quick Guide](#toc0_)
---

#### <a id='toc1_5_1_1_'></a>[🔹 What is a Mortgage?](#toc0_)

A **mortgage** is a loan used to purchase real estate, where the property itself acts as collateral. The borrower agrees to pay back the loan over a fixed term through monthly payments.

---

#### <a id='toc1_5_1_2_'></a>[ 🔹 Fixed-Interest Mortgage](#toc0_)

In a **fixed-rate mortgage**, the interest rate remains constant over the entire loan term. The monthly payment includes:

- **Principal** – the original loan amount
- **Interest** – the cost of borrowing

#### <a id='toc1_5_1_3_'></a>[ 🧮 Monthly Payment Formula:](#toc0_)

$$
M = P \cdot \frac{r(1 + r)^n}{(1 + r)^n - 1}
$$

Where:
- $ M $ = monthly payment  
- $ P $ = loan principal  
- $ r $ = monthly interest rate = annual rate ÷ 12  
- $ n $ = total number of payments (months)

---

#### <a id='toc1_5_1_4_'></a>[ 📘 Example:](#toc0_)

Loan Amount = ₹300,000  
Annual Interest = 12% → Monthly rate = 0.12 / 12 = 0.01  
Term = 5 years = 60 months

$$
M = 300,000 \cdot \frac{0.005(1 + 0.005)^{360}}{(1 + 0.005)^{360} - 1} \approx ₹1,798.65
$$

---

#### <a id='toc1_5_1_5_'></a>[ 🔹 Mortgage Points](#toc0_)

**Points** are prepaid interest paid at closing to reduce the mortgage interest rate.

- **1 point = 1% of the loan amount**
- Lowers your monthly payment by reducing your interest rate
- Makes sense if you plan to **stay in the home long-term**

#### <a id='toc1_5_1_6_'></a>[ 💡 Tradeoff:](#toc0_)
- Pay more upfront → Save more over time

---

#### <a id='toc1_5_1_7_'></a>[ 📘 Example:](#toc0_)

Loan = ₹200,000  
1 point = 1% of ₹200,000 = ₹2,000  
Reduces rate from 6% to 5.75%, for example

---

#### <a id='toc1_5_1_8_'></a>[ 🔹 Two-Rate (Dual-Rate) Mortgage](#toc0_)

This mortgage starts with a **low fixed interest rate** for an initial period and then **switches to a higher variable or standard rate**.

#### <a id='toc1_5_1_9_'></a>[ 🔄 Structure:](#toc0_)
- Years 0–3: Fixed rate (e.g., 4.0%)
- Year 4 onwards: Variable rate (e.g., 6.5%)

#### <a id='toc1_5_1_10_'></a>[ 🧠 Use Cases:](#toc0_)
- Beneficial if borrower plans to **sell or refinance** early
- Risk: Future payments may rise significantly

---

#### <a id='toc1_5_1_11_'></a>[ 🔹 Amortization (How Payments Are Applied)](#toc0_)

Every monthly payment splits into:

- **Interest** = Outstanding Principal × Monthly Interest Rate  
- **Principal** = Remaining part of the payment that reduces the loan

Over time:
- **Interest portion decreases**
- **Principal portion increases**

---

#### <a id='toc1_5_1_12_'></a>[ 📘 Example (Month 1):](#toc0_)

Loan = \$300,000  
Monthly Rate = 0.005  
Monthly Payment = \$1,798.65

- Interest = \$300,000 × 0.005 = \$1,500  
- Principal = \$1,798.65 - \$1,500 = \$298.65  
- New balance = \$300,000 - \$298.65 = \$299,701.35

🧪 Mortgage class

```python
def find_payment(loan:float, rate:float, months:int):
    """ return monthly payment for a mortgage of size loan at a monthly rate of rate for months

    Args:
        loan (float): _description_
        rate (float): _description_
        months (int): _description_
    """
    
    return loan*((rate*(1+rate))**months) / ((1+rate)**months - 1)

class Mortgage(object):
    def __init__(self, loan:float, ann_rate:float, months:int):
        self._loan = loan 
        self._rate = ann_rate
        self._months = months 
        self._paid = [0.0]
        self._outstanding = [loan]
        self._payment = find_payment(self._loan, self._rate,self._months)
        self._legend = None # description of mortgage
    
    def make_payment(self):
        self._paid.append(self._payment)
        reduction = self._payment - self._outstanding[-1]*self._rate
        self._outstanding.append(self._outstanding[-1] - reduction)
        
    def get_total_paid(self):
        """Return total amount paid so far
        """
        try:
            return sum(self._paid)
        except ValueError:
            return f"Non number found"
    def __str__(self):
        return self._legend
```
