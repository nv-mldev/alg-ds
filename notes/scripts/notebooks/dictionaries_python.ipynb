# Dictionaries

- Dictionaries are unordered collections of key-value pairs before Python 3.7  
- Dictionaries are ordered collections of key-value pairs in Python 3.7 and later  
- Dictionaries can contain items of different types  
- Dictionaries are mutable, meaning they can be changed after creation  
- Dictionaries are created using curly braces `{}`  


```
# Create an empty dictionary
my_dict = {}

# Create a dictionary with items
my_dict = {'name': 'David', 'age': 32}
my_another_dict = {'name': 'Alice', 'age': 28, 'is_student': True}
```

## Dictionary Methods in Python (Part 1)

| Method      | Use                       | Explanation                                           |
|-------------|---------------------------|-------------------------------------------------------|
| `clear`     | `a_dict.clear()`          | Removes all items from the dictionary                |
| `copy`      | `a_dict.copy()`           | Returns a shallow copy of the dictionary             |
| `get`       | `a_dict.get(key, default)`| Returns the value for `key` if it exists             |
| `items`     | `a_dict.items()`          | Returns key-value pairs of all items in the dictionary |
| `keys`      | `a_dict.keys()`           | Returns a view of all keys in the dictionary         |


## Dictionary Methods in Python (Part 2)

| Method        | Use                                | Explanation                                               |
|---------------|------------------------------------|-----------------------------------------------------------|
| `pop`         | `a_dict.pop(key)`                  | Removes the item with `key` and returns its value         |
| `popitem`     | `a_dict.popitem()`                 | Removes and returns an arbitrary item from the dictionary |
| `setdefault`  | `a_dict.setdefault(key, default)`  | Returns value if key exists, else sets it to `default`    |


## Dictionary Methods in Python (Part 3)

| Method      | Use                            | Explanation                                           |
|-------------|---------------------------------|-------------------------------------------------------|
| `update`    | `a_dict.update(other_dict)`     | Updates the dictionary with items from `other_dict`   |
| `values`    | `a_dict.values()`               | Returns a view of all values in the dictionary        |


```
# Create two dictionaries
dict1 = {'name': 'David', 'age': 32}
dict2 = {'name': 'Alice', 'age': 28, 'is_student': True}

# Update a dictionary
dict1.update(dict2)
print(dict1)

# Get a value from a dictionary
print(dict1.get('name'))

# Remove an item from a dictionary
dict1.pop('age')
```

```
# Create a dictionary
my_dict = {'name': 'David', 'age': 32}

# Add a new item to the dictionary 
my_dict['is_student'] = True
print(my_dict)

# Remove an item from the dictionary
del my_dict['age']
print(my_dict)
```

```
# Create a dictionary
my_dict = {'name': 'David', 'age': 32}

# Check if a key exists in the dictionary
print('name' in my_dict)
print('is_student' in my_dict)

# Get the number of items in the dictionary
print(len(my_dict))
```
