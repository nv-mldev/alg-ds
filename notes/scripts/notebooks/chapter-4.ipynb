### <a id='toc1_2_7_'></a>[🧠 Mutability of Lists in Python](#toc0_)
* Lists in Python are mutable, meaning their contents can be changed after creation.
* Example: Modifying a list through a reference.

```python
# Create a list
a = [1]

# Assign the list to another variable
b = a

# Modify the list through the original variable
a[0] = 5

# Print the modified list through the reference
print(b)  # Output: [5]
```

* Many operators can be used to create objects of immutable types, and variables can be bound to objects of these types. 
* But objects of immutable types cannot be modified after they are created.
* Objects of mutable types can be modified after they are created

**🤔 What is happening ?**

* We have here is called aliasing. There are two distinct paths to the same list object. 

* One path is through the variable Techs and the other through the first element of the list object to which
Univs is bound. 

* We can mutate the object via either path, and the effect of the mutation will be visible through both paths. 

* This can be convenient, but it can also be treacherous. 

* Unintentional aliasing leads to programming errors that are often enormously hard to track down

```python
L1 = [[]]*2
L2 = [[], []]
print(id(L1)== id(L2))
for i in range(len(L1)):
    L1[i].append(i)
    L2[i].append(i)
print('L1 =', L1, 'but', 'L2 =', L2)
```

**🤔 What is happening** 
* first assignment statement creates a list with two elements, each of which is the same object
* the second assignment statement creates a list with two different objects, each of which is initially equal to an empty list 

```python
L = [1, 2, 3]
L.append(L)
print(L is L[-1])
```

**🤔 What is happening ?**

* this creates a recrusive list or infinite list

```python
def append_val(val, list_1 = []):
    list_1.append(val)
    return list_1
print(append_val(3))
print(append_val(4))
```

**🤔 What is happening ?**

* `list_1` is the default and if not given a list specifically it is bounded to the empty list that created at the first time function is called 
* every other call with out a list will just mutate the old list

### <a id='toc1_2_8_'></a>[🚀 Cloning Lists](#toc0_)



```python
def remove_dups(L1, L2):
    for e1 in L1:
        if e1 in L2:
            L1.remove(e1)
L1 = [1,2,3,4]
L2 = [1,2,5,6]
remove_dups(L1, L2)
print('L1 =', L1)
```

**🤔 What is happening?**
* During a for loop, Python keeps track of where it is in the list using an internal counter that is incremented at the end of each
iteration. When the value of the counter reaches the current length of the list, the loop terminates.

* In this case the hidden counter starts out at 0, discovers that `L1[0]` is in `L2`, and removes it—reducing the length of
`L1` to 3. 

* The counter is then incremented to 1, and the code proceeds to check if the value of `L1[1]` is in `L2`. Notice that this is not the
original value of `L1[1]` (i.e., 2), but rather the current value of `L1[1]`(i.e., 3).

```python
def remove_dups(L1, L2):
    for e1 in L1[:]:
        if e1 in L2:
            L1.remove(e1)
L1 = [1,2,3,4]
L2 = [1,2,5,6]
remove_dups(L1, L2)
print('L1 =', L1)
```

**🤔 What is happening?**
* creates a clone using `L1[:]` or use `L1.copy()`

* Both slicing and copy perform what is known as a **shallow copy**

* **shallow copy** = Copy the container, but not the contents.It creates a new list, but the elements inside the list are still references to the same original objects.


```python
import copy
L = [2]
L1 = [L]
L2 = L1[:]
L3 = copy.deepcopy(L1)
L.append(3)
print(f'L1 = {L1}, L2 = {L2}, L3 = {L3}')
```

**🤔 What is happening?**
* If the list to be copied contains mutable objects that you also want to copy, import the standard library module `copy` and use the
function `copy.deepcopy` to make a **deep copy**

* The method deepcopy creates a new list and then inserts copies of the objects in the list to be copied into the new list

```python
L1 = [2]
L2 = [[L1]]
L3 = copy.deepcopy(L2)
L1.append(3)
print(f'L1 = {L1}, L2 = {L2}, L3 = {L3}')
```

```python
L1 = [2]
L2 = [L1, L1]
L3 = copy.deepcopy(L2)
L3[0].append(3)
print(L3)
```

**🤔 What is happening?**

* `L3` will just create a copy of `L2` but creates a single copy of `L1` and uses it

### <a id='toc1_2_9_'></a>[⛓️List Comprehension](#toc0_)

* List comprehension provides a concise way to apply an operation to the sequence values provided by iterating over an iterable value

* It creates a new list in which each element is the result of applying a given operation to a value from an iterable

* expression : 
  *[expr for elem in iterable if test]*

```python
# List comprehension
print(f'List Comprehension : {[e**2 for e in range(10)]}')

# List comprehension with condition
print(f'List comprehension with condition : {[e**2 for e in range(10) if e % 2 == 0]}')

# Nested list comprehension
print(f'Nested list comprehension : {[[e**2 for e in range(10)] for i in range(3)]}')

# Nested list comprehension with condition
print(f'Nested list comprehension with condition : {[[e**2 for e in range(10) if e % 2 == 0] for i in range(3)]}')

# List comprehension with multiple loops
print(f'List comprehension with multiple loops : {[(x, y) for x in range(3) for y in range(2)]}')

# List comprehension with multiple loops and condition
print(f'List comprehension with multiple loops and condition: {[(x, y) for x in range(3) for y in range(2) if x + y > 2]}')
```

**🧪 use nested list comprehensions to generate a list of all prime numbers less than 100**

```python
prime_numbers= [x for x in range(2,100) if all(x % y !=0 for y in range(2,x)) ] 
print(prime_numbers)
```

### <a id='toc1_5_3_'></a>[🔁 Mutability of Sets in Python](#toc0_)

- **Sets themselves are mutable**
  - Add with `add()`
  - Remove with `remove()` or `discard()`
  - Combine with `update()` or `union()`

- **Set elements must be immutable**
  - Cannot include lists or dicts
  - Can include integers, strings, tuples

- **Why the confusion?**
  - Some sources say 'immutable' due to element restrictions
  - Sets rely on hashable (unchangeable) elements

```python
# Create an empty set
my_set = set()
# Create a set with items
my_set = {1, 2, 3, 4, 5}
my_another_set = {'a', 'b', 'c', False, 3, 0.11}
```

```python
set1 = {1, 2, 3, 4, 5}  
set2 = {4, 5, 6, 7, 8}

# Union
union_set = set1.union(set2)
print(union_set)

# Difference
difference_set = set1.difference(set2)
print(difference_set)

# Intersection
intersection_set = set1.intersection(set2)
print(intersection_set)
```

#### <a id='toc1_6_1_2_'></a>[Dictionary Methods](#toc0_)

| Method        | Use                                | Explanation                                               |
|---------------|------------------------------------|-----------------------------------------------------------|
| `pop`         | `a_dict.pop(key)`                  | Removes the item with `key` and returns its value         |
| `popitem`     | `a_dict.popitem()`                 | Removes and returns an arbitrary item from the dictionary |
| `setdefault`  | `a_dict.setdefault(key, default)`  | Returns value if key exists, else sets it to `default`    |


#### <a id='toc1_6_1_3_'></a>[Dictionary Methods](#toc0_)

| Method      | Use                            | Explanation                                           |
|-------------|---------------------------------|-------------------------------------------------------|
| `update`    | `a_dict.update(other_dict)`     | Updates the dictionary with items from `other_dict`   |
| `values`    | `a_dict.values()`               | Returns a view of all values in the dictionary        |


```python
# Create two dictionaries
dict1 = {'name': 'David', 'age': 32}
dict2 = {'name': 'Alice', 'age': 28, 'is_student': True}

# Update a dictionary
dict1.update(dict2)
print(dict1)

# Get a value from a dictionary
print(dict1.get('name'))

# Remove an item from a dictionary
dict1.pop('age')
```

```python
# Create a dictionary
my_dict = {'name': 'David', 'age': 32}

# Add a new item to the dictionary 
my_dict['is_student'] = True
print(my_dict)

# Remove an item from the dictionary
del my_dict['age']
print(my_dict)
```
